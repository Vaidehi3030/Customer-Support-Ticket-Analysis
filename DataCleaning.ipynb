#Working on Dataset

```python
import pandas as pd
import numpy as np
```

```python
#Load Dataset

df = pd.read_csv("customer_support_tickets.csv")
df.head()
```

```python
#Size of Database

df.shape
```

```python
#Structure

df.info()
```

```python
#Check Missing Values

df.isnull().sum()
```

```python
#Convert Date Columns

df['Created_Date'] = pd.to_datetime(df['Created_Date'], format='%d-%m-%Y')
df['Resolved_Date'] = pd.to_datetime(df['Resolved_Date'], format='%d-%m-%Y')
```

```python
df.info()
```

```python
#Calculate Resolution Time

df['Resolution_Time_Hours'] = (df['Resolved_Date'] - df['Created_Date']).dt.total_seconds() / 3600
df.head()
```

```python
#Average Res Time is 11.52.
#50% of tickets are resolved on the same day which means actively working.
#Remaining are resolved within 24 hours of raising ticket.

df['Resolution_Time_Hours'].describe()
```

#Basic Data Analytics

```python
#Total Tickets

total_tickets = df.shape[0]
print(total_tickets)
```

```python
#Tickets by Region
#Most tickets are raised in the North Region

df['Region'].value_counts()
```

```python
#Most Common Issue
#Login issue is common and least commmon is Locked Account

df['Issue_Type'].value_counts()
```

```python
#Average Resolution Time
#Approximately 12 hours

df['Resolution_Time_Hours'].mean()
```

```python
#Agent Performance
#Neha & Sneha have delayed solving most tickets.

df.groupby("Agent")['Resolution_Time_Hours'].mean().sort_values(ascending=False)
```

```python
#Customer Satisfaction
#Average satisfaction is from Neha and Rahul

df.groupby("Agent")['Customer_Satisfaction'].mean()
```
